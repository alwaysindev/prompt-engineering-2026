# Demo 1 — Los tokenizadores y el mito del español

**Se graba en el capítulo 2, ~30 segundos en pantalla.**

Orden de ejecución: 1 → 2 → 3 → 4. Ejecuta todo **antes** de grabar para que los modelos BPE
queden cacheados y no haya esperas de descarga delante de la cámara.

Requisito: `pip install tiktoken`

```python
import tiktoken

# Los tres tokenizadores de la historia reciente
TOKENIZADORES = {
    "r50k_base  (GPT-3)":   tiktoken.get_encoding("r50k_base"),
    "cl100k_base (GPT-4)":  tiktoken.get_encoding("cl100k_base"),
    "o200k_base  (actual)": tiktoken.get_encoding("o200k_base"),
}
print("Listos:", *TOKENIZADORES, sep="\n  ")
```

## 1. Qué es un token — las piezas de Lego

Esta celda es el apoyo visual de la analogía. Se ve que las palabras comunes son **una pieza**
y las raras se parten en varias.

```python
enc = TOKENIZADORES["o200k_base  (actual)"]

for palabra in ["casa", "tokenización", "electroencefalografista", "ladrillos"]:
    piezas = [enc.decode([t]) for t in enc.encode(palabra)]
    print(f"{palabra:28} → {len(piezas)} token(s):  " + " | ".join(piezas))
```

## 2. El experimento: el mismo párrafo en español y en inglés

⚠️ Cambia el párrafo si quieres, pero **mantén los dos textos equivalentes**: si uno dice más
cosas que el otro, la comparación no vale nada.

```python
ES = ("El modelo no lee palabras, lee tokens. Un token es un trozo de palabra: a veces la "
      "palabra entera, a veces una sílaba, a veces dos letras sueltas. Da igual lo bien que "
      "escribas: si no entiendes cómo se parte tu texto, no entiendes lo que estás pagando "
      "ni por qué el modelo se equivoca contando letras.")

EN = ("The model does not read words, it reads tokens. A token is a chunk of a word: sometimes "
      "the whole word, sometimes a syllable, sometimes two loose letters. It does not matter how "
      "well you write: if you do not understand how your text is split, you do not understand "
      "what you are paying for, nor why the model gets letter counting wrong.")

print(f"Español: {len(ES.split())} palabras\nInglés:  {len(EN.split())} palabras")
```

## 3. La medición — este es el número que sale en pantalla

```python
print(f"{'TOKENIZADOR':22} {'ES':>6} {'EN':>6} {'SOBRECOSTE DEL ESPAÑOL':>24}")
print("-" * 62)

for nombre, tok in TOKENIZADORES.items():
    n_es, n_en = len(tok.encode(ES)), len(tok.encode(EN))
    extra = (n_es / n_en - 1) * 100
    print(f"{nombre:22} {n_es:>6} {n_en:>6} {extra:>23.1f}%")

print("\n→ El consejo de escribir en inglés para ahorrar era de la primera fila.")
```

## 4. Por qué ha cambiado (opcional, si sobra tiempo)

La misma palabra, partida por cada generación de tokenizador. Se ve la caja de piezas
llenándose de español.

```python
for palabra in ["ladrillos", "aprendizaje", "escribiendo"]:
    print(f"\n{palabra}")
    for nombre, tok in TOKENIZADORES.items():
        piezas = [tok.decode([t]) for t in tok.encode(palabra)]
        print(f"   {nombre:22} {len(piezas)} → " + " | ".join(piezas))
```
